# Extract Instagram message HTML to JSON

1. Reads `data/message_1.html` and writes `data/message_1.json`
2. Chunks messages by day (day starts at **5 AM**) into `data/chats/<year>-<month>-<day>/raw_chat.json` (e.g. `23-7-3/raw_chat.json`, date label `23/7/3`)

```python
from pathlib import Path

from extract_message_html import extract_messages, write_json

root = Path("..").resolve()
input_path = root / "data" / "message_1.html"
output_path = root / "data" / "message_1.json"

data = extract_messages(input_path)
write_json(data, output_path)

print(f"Thread: {data['thread_title']}")
print(f"Participants: {', '.join(data['participants'])}")
print(f"Messages: {data['message_count']}")
print(f"Saved to: {output_path}")
data["messages"][:3]
```

```python
from chunk_by_day import chunk_messages_by_day

chunks_root = root / "data" / "chats"
written = chunk_messages_by_day(data, chunks_root)

print(f"Created {len(written)} day folders under {chunks_root}")
written[:5]
```
